```python
# Validate YOLO11n model
```

```python
import os
import numpy as np
import pandas as pd
from ultralytics import YOLO
from ultralytics.utils.torch_utils import get_flops

MODEL_LABEL  = "YOLO-11"                 # name used in the CSV 'model' column
RUN_NAME     = "Yolo11-1024seed{seed}"      # folder pattern; MUST contain {seed}
seeds        = [0, 1, 2]
data_yaml    = "/content/drive/MyDrive/740-image/data_config.yaml"
weights_root = "/content/drive/MyDrive/740-image/Result/Ablation"
CSV_DIR      = "/content/drive/MyDrive/740-image/Result/CSV/YOLO11"

VAL_ARGS = dict(
    data         = data_yaml,
    split        = "val",
    imgsz        = 1024,
    batch        = 16,
    device       = "cuda:0",
    conf         = 0.001,
    iou          = 0.7,
    agnostic_nms = True,
    max_det      = 500,
)

assert "{seed}" in RUN_NAME, "RUN_NAME must contain the {seed} placeholder"
os.makedirs(CSV_DIR, exist_ok=True)

def append_csv(df, path):
    """Append rows, writing the header only when the file is new.
    Re-running the same model duplicates its rows; delete the CSV to start clean."""
    df.to_csv(path, mode="a", header=not os.path.exists(path), index=False)

overall_rows, class_rows, size_mb = [], [], []
n_params = n_gflops = None

# ---- one validation per seed ------------------------------------------------
for seed in seeds:
    weights = f"{weights_root}/{RUN_NAME.format(seed=seed)}/weights/best.pt"
    size_mb.append(os.path.getsize(weights) / 1e6)

    model = YOLO(weights)

    if n_params is None:                       # identical across seeds; record once
        n_params = sum(p.numel() for p in model.model.parameters())
        n_gflops = get_flops(model.model, VAL_ARGS["imgsz"])

    m = model.val(**VAL_ARGS)                  # runs for EVERY seed

    overall_rows.append(dict(model=MODEL_LABEL, seed=seed, imgsz=VAL_ARGS["imgsz"],
                             precision=float(m.box.mp), recall=float(m.box.mr),
                             mAP50=float(m.box.map50), mAP50_95=float(m.box.map)))

    for i, cls_idx in enumerate(m.box.ap_class_index):
        class_rows.append(dict(model=MODEL_LABEL, seed=seed, cls=m.names[cls_idx],
                               precision=float(m.box.p[i]), recall=float(m.box.r[i]),
                               mAP50=float(m.box.ap50[i]), mAP50_95=float(m.box.ap[i])))

    print(f"seed {seed}: P {m.box.mp:.4f}  R {m.box.mr:.4f}  "
          f"mAP50 {m.box.map50:.4f}  mAP50-95 {m.box.map:.4f}  size {size_mb[-1]:.2f} MB")

# ---- assemble ---------------------------------------------------------------
overall_df = pd.DataFrame(overall_rows)
class_df   = pd.DataFrame(class_rows)
cost_df    = pd.DataFrame([dict(model=MODEL_LABEL, params_M=round(n_params / 1e6, 3),
                                GFLOPs=round(float(n_gflops), 2),
                                size_MB=round(float(np.mean(size_mb)), 2),
                                n_seeds=len(seeds))])

print("\nRows to be written:")
print(overall_df.round(4).to_string(index=False))
assert overall_df["mAP50_95"].nunique() > 1, \
    "all seeds gave identical results - check that model.val() runs inside the loop"

append_csv(overall_df, f"{CSV_DIR}/per_seed_overall.csv")
append_csv(class_df,   f"{CSV_DIR}/per_seed_class.csv")
append_csv(cost_df,    f"{CSV_DIR}/model_cost.csv")

# ---- summaries --------------------------------------------------------------
METRICS = ["precision", "recall", "mAP50", "mAP50_95"]

def summarise(df, keys):
    g = df.groupby(keys)[METRICS]
    out = g.mean().round(4).astype(str) + " \u00b1 " + g.std(ddof=1).round(4).astype(str)
    return out.reset_index()

n = len(seeds)
print(f"\nPer-class (mean \u00b1 sample std across {n} seeds):")
print(summarise(class_df, ["cls"]).to_string(index=False))
print(f"\nOverall (mean \u00b1 sample std across {n} seeds):")
print(summarise(overall_df, ["model"]).to_string(index=False))
print("\nModel cost:")
print(cost_df.to_string(index=False))
print(f"\nAppended to {CSV_DIR}/")

```
